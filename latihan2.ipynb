### Algoritma Grover ###
Algoritma kuantum utk pencarian dalam database tak terurut dengan peningkatan kecepatan kuadrat dibandingkan algoritma klasik 

### Solusi Klasik ###

```python
# data tidak beraturan 
data = [1,2,5,6,4,7,0,9,3,8]

# angka target
target = 6

# oracle pada komputasi klasik
def classical_oracle(inp):
    if inp == target:
        response = 1
    else:
        response = 0
    return response

# metode search pada komputasi klasik
for i, val in enumerate(data):
    if classical_oracle(val) == 1:
        print(f"Angka {val} ditemukan!")
        print(f"membutuhkan {i+1} query") #query adalah baris perintah yg digunakan utk mendapatkan informasi dari database
        break
```
