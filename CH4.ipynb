```python
import matplotlib.pyplot as plt
import numpy as np

# Data ORCA scan CH4 (CASSCF/STO-3G)
jarak = np.linspace(0.50, 2.50, 21)
energi = [
    -38.709011827266,
    -39.202617107726,
    -39.480720810134,
    -39.634211167725,
    -39.714226277948,
    -39.750553203875,
    -39.760717981202,
    -39.755186429032,
    -39.740429830294,
    -39.720628077791,
    -39.698568902006,
    -39.676139790348,
    -39.654622551064,
    -39.634874321787,
    -39.617429565835,
    -39.602553773675,
    -39.590280484780,
    -39.580456147464,
    -39.572799137118,
    -39.566964564743,
    -39.562600203282
]
# Cari minimum
idx_min = energi.index(min(energi))
r_min   = jarak[idx_min]
e_min   = energi[idx_min]

fig, ax = plt.subplots(figsize=(9, 6))

ax.plot(jarak, energi, 'o-', color='steelblue', linewidth=2,
        markersize=7, markerfacecolor='white', markeredgewidth=2,
        label='CASSCF(4,4)/STO-3G (Klasik)')

ax.plot(r_min, e_min, 'ro', markersize=10, zorder=5)

ax.annotate(f'R_eq = {r_min:.2f} Å\nE = {e_min:.5f} Ha',
            xy=(r_min, e_min),
            xytext=(r_min+0.25, e_min+0.07),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
            fontsize=10,
            bbox=dict(boxstyle='round,pad=0.4',
                      facecolor='white',
                      edgecolor='black',
                      linewidth=1))

ax.axvline(x=1.089, color='orange', linestyle='--',
           linewidth=1.5, alpha=0.8, label='Eksperimen C-H = 1.089 Å')

ax.set_xlabel('Panjang Ikatan C-H (Å)', fontsize=13)
ax.set_ylabel('Energi (Hartree)', fontsize=13)
ax.set_title('Kurva Energi Potensial CH₄\nCASSCF(4,4)/STO-3G', fontsize=13)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('PES_CH4_klasik.png', dpi=150, bbox_inches='tight')
plt.show()
print(f"R optimal = {r_min:.2f} Å")
print(f"E minimum = {e_min:.6f} Eh")
```
