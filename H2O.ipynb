```python
import matplotlib.pyplot as plt
import numpy as np

# Data ORCA scan H2O (M062X/STO-3G)
jarak = np.linspace(0.50, 2.50, 21)
energi = [
    -74.053978499751,
    -74.566450042709,
    -74.822303169472,
    -74.946597615627,
    -75.001236876913,
    -75.017713423613,
    -75.013153571026,
    -74.997671413200,
    -74.977437153115,
    -74.956116909891,
    -74.935794929920,
    -74.917611530065,
    -74.902132517010,
    -74.889524115776,
    -74.879651587403,
    -74.872178137316,
    -74.866672426364,
    -74.862699269585,
    -74.859875441876,
    -74.857891280296,
    -74.856509871211
]
# Cari minimum
idx_min = energi.index(min(energi))
r_min   = jarak[idx_min]
e_min   = energi[idx_min]

fig, ax = plt.subplots(figsize=(9, 6))

ax.plot(jarak, energi, 'o-', color='steelblue', linewidth=2,
        markersize=7, markerfacecolor='white', markeredgewidth=2,
        label='CASSCF/STO-3G (Klasik)')

ax.plot(r_min, e_min, 'ro', markersize=10, zorder=5)

ax.annotate(f'R_eq = {r_min:.2f} Å\nE = {e_min:.5f} Ha',
            xy=(r_min, e_min),
            xytext=(r_min+0.25, e_min+0.07),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
            fontsize=10,
            bbox=dict(boxstyle='round,pad=0.4',
                      facecolor='white',
                      edgecolor='black',
                      linewidth=1))

ax.axvline(x=0.9584, color='orange', linestyle='--',
           linewidth=1.5, alpha=0.8, label='Eksperimen O-H = 0.9584 Å')

ax.set_xlabel('Panjang Ikatan O-H (Å)', fontsize=13)
ax.set_ylabel('Energi (Hartree)', fontsize=13)
ax.set_title('Kurva Energi Potensial H₂O\nCASSCF/STO-3G', fontsize=13)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('PES_H2O_klasik.png', dpi=150, bbox_inches='tight')
plt.show()
print(f"R optimal = {r_min:.2f} Å")
print(f"E minimum = {e_min:.6f} Eh") 
```
