```python
import matplotlib.pyplot as plt
import numpy as np

# Data ORCA scan
jarak = [0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90, 1.00, 1.10, 1.20,
         1.30, 1.40, 1.50, 1.60, 1.70, 1.80, 1.90, 2.00, 2.10, 2.20,
         2.30, 2.40, 2.50]
energi = [-0.690922389, -0.973734451, -1.097499757, -1.149977566,
          -1.167258203, -1.166302711, -1.155688885, -1.140011607,
          -1.121902427, -1.102940830, -1.084083836, -1.065905467,
          -1.048626184, -1.032477577, -1.017499661, -1.003485586,
          -0.990736547, -0.978916036, -0.968025243, -0.958153548,
          -0.949081467, -0.940760775, -0.933149921]

# Cari minimum
idx_min = energi.index(min(energi))
r_min   = jarak[idx_min]
e_min   = energi[idx_min]

fig, ax = plt.subplots(figsize=(9, 6))

ax.plot(jarak, energi, 'o-', color='steelblue', linewidth=2,
        markersize=7, markerfacecolor='white', markeredgewidth=2,
        label='M062X/6-311G(d,p)')

ax.plot(r_min, e_min, 'ro', markersize=10, zorder=5)

# Kotak anotasi seperti VQE
ax.annotate(f'R_eq = {r_min:.2f} Å\nE = {e_min:.5f} Ha',
            xy=(r_min, e_min),
            xytext=(r_min+0.25, e_min+0.07),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
            fontsize=10,
            bbox=dict(boxstyle='round,pad=0.4',
                      facecolor='white',
                      edgecolor='black',
                      linewidth=1))

ax.axvline(x=0.7414, color='orange', linestyle='--',
           linewidth=1.5, alpha=0.8, label='Eksperimen = 0.7414 Å')

ax.set_xlabel('Panjang Ikatan H-H (Å)', fontsize=13)
ax.set_ylabel('Energi (Hartree)', fontsize=13)
ax.set_title('Kurva Energi Potensial H₂\nM062X/6-311G(d,p)', fontsize=13)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('PES_H2_orca.png', dpi=150, bbox_inches='tight')
plt.show()
print(f"R optimal = {r_min} Å")
print(f"E minimum = {e_min:.6f} Eh")
```
