```python
import matplotlib.pyplot as plt
import numpy as np

# Data ORCA scan (M062X/STO-3G)
jarak = [0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90, 1.00, 1.10, 1.20,
         1.30, 1.40, 1.50, 1.60, 1.70, 1.80, 1.90, 2.00, 2.10, 2.20,
         2.30, 2.40, 2.50]
energi = [-0.627720651, -0.938565407, -1.078400508, -1.137028444,
          -1.154229084, -1.149981807, -1.133944791, -1.111569809,
          -1.085960171, -1.059050211, -1.031722371, -1.004921827,
          -0.979106675, -0.954816102, -0.932222646, -0.911206563,
          -0.891919820, -0.874270812, -0.858277883, -0.843932841,
          -0.831028409, -0.819554551, -0.809406369]

idx_min = energi.index(min(energi))
r_min   = jarak[idx_min]
e_min   = energi[idx_min]

fig, ax = plt.subplots(figsize=(9, 6))

ax.plot(jarak, energi, 'o-', color='steelblue', linewidth=2,
        markersize=7, markerfacecolor='white', markeredgewidth=2,
        label='M062X/STO-3G')

ax.plot(r_min, e_min, 'ko', markersize=10, zorder=5)
ax.annotate(f'R_eq = {r_min:.2f} Å\nE = {e_min:.5f} Ha',
            xy=(r_min, e_min), xytext=(r_min+0.2, e_min+0.06),
            arrowprops=dict(arrowstyle='->', color='black'),
            fontsize=10,
            bbox=dict(boxstyle='round,pad=0.4', facecolor='white',
                      edgecolor='black', linewidth=1))

ax.axvline(x=0.7414, color='orange', linestyle='--',
           linewidth=1.5, alpha=0.8, label='Eksperimen = 0.7414 Å')

ax.set_xlabel('Panjang Ikatan H-H (Å)', fontsize=13)
ax.set_ylabel('Energi (Hartree)', fontsize=13)
ax.set_title('Kurva Energi Potensial H₂\nM062X/STO-3G (ORCA)', fontsize=13)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('PES_H2_orca_sto3g.png', dpi=150, bbox_inches='tight')
plt.show()
print(f"R optimal = {r_min} Å")
print(f"E minimum = {e_min:.6f} Eh")
```
